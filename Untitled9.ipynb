```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mashlyn/online-retail-ii-uci")

print("Path to dataset files:", path)
```

```python
import os

os.listdir(path)
```

```python
import pandas as pd

file_path = path + "/online_retail_II.csv"
df = pd.read_csv(file_path)

df.head()
```

```python
df.head()
df.shape
df.info()
df.describe()

original_df = df.copy()
```

```python
# Check missing values
df.isnull().sum()
```

```python
# Drop rows without customer ID
df = df.dropna(subset=['Customer ID'])

# Fill description
df['Description'] = df['Description'].fillna('Unknown')
```

```python
# Drop duplicates
df.duplicated().sum()
df = df.drop_duplicates()
```

```python
# Convert values to the same data type
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['Customer ID'] = df['Customer ID'].astype(int)
```

```python
# Remove negative quantities
df = df[df['Quantity'] > 0]
```

```python
# Remove negative prices
df = df[df['Price'] > 0]
```

```python
# Use IQR to identify outliers and remove them

Q1 = df['Quantity'].quantile(0.25)
Q3 = df['Quantity'].quantile(0.75)
IQR = Q3 - Q1

df = df[(df['Quantity'] >= Q1 - 1.5*IQR) &
        (df['Quantity'] <= Q3 + 1.5*IQR)]
```

```python
# Standardization

# Clean column names
df.columns = df.columns.str.lower().str.replace(" ", "_")

# Clean text
df['country'] = df['country'].str.lower().str.strip()
df['description'] = df['description'].str.lower().str.strip()
```

```python
print("Original shape:", original_df.shape)
print("Cleaned shape:", df.shape)

print("Missing values BEFORE:")
print(original_df.isnull().sum())

print("\nMissing values AFTER:")
print(df.isnull().sum())

print("Duplicates BEFORE:", original_df.duplicated().sum())
print("Duplicates AFTER:", df.duplicated().sum())
```

```python
df.to_csv("cleaned_online_retail.csv", index=False)
```
